```python
import pandas as pd

# 1) Extract region, times, and direction from the path
parts = df["npz_path"].str.extract(
    r"/HV_HH/(?P<region>region-[^/]+)/\d{4}/\d{2}/\d{2}/"
    r"(?P<t0>\d{8}T\d{4})__(?P<t1>\d{8}T\d{4})(?:_backward)?_(?P<dir>past|future)\.npz$"
)
print(parts)
df = df.join(parts)

df["t0"] = pd.to_datetime(df["t0"], format="%Y%m%dT%H%M", errors="coerce")
df["t1"] = pd.to_datetime(df["t1"], format="%Y%m%dT%H%M", errors="coerce")
df["dir"] = df["dir"].astype("category")

# 2) Middle timestamp: past uses t1, future uses t0
middle_time = df["t1"].where(df["dir"].eq("past"), df["t0"])

# 3) Composite key prevents cross-region pairing
df["pair_key"] = list(zip(df["region"], middle_time))

# 4) Check pairing: each key should have exactly one past + one future
pair_counts = (
    df.groupby(["pair_key", "dir"])
      .size()
      .unstack(fill_value=0)
)

problems = pair_counts[(pair_counts.get("past", 0) != 1) | (pair_counts.get("future", 0) != 1)]

print("Total unique pair_keys:", len(pair_counts))
print("Problematic keys:", len(problems))

# Optional: see distribution (ideally only (1,1))
print(pair_counts.value_counts().head(10))
```

```python
# split past and future
past_df = df[df["dir"] == "past"].set_index("pair_key")
future_df = df[df["dir"] == "future"].set_index("pair_key")

# join into one row per pair
paired = (
    past_df[["npz_path", "n_pm_good", "n_pm_total", "n_ft"]].add_prefix("past_")
    .join(
        future_df[["npz_path", "n_pm_good", "n_pm_total", "n_ft"]].add_prefix("future_"),
        how="inner"
    )
)

print("Total paired samples:", len(paired))
```

```python
paired = paired[
    (paired['past_n_pm_good'] > 399) &
    (paired['future_n_pm_good'] > 399)
]
```

```python
out_path = "paired_HV_HH_min400PM.csv"

paired.reset_index().to_csv(out_path, index=False)

print(f"Saved {len(paired)} rows to {out_path}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import BoundaryNorm

# 1) Extract all middle times from paired.index
dates = pd.to_datetime([k[1] for k in paired.index], errors="coerce")
dates = dates[~pd.isna(dates)]  # drop NaT if any
dates
```

```python
plt.figure(figsize=(12, 3))
plt.plot(dates, np.ones(len(dates)), "|")
plt.yticks([])
plt.xlabel("Date")
plt.title("Timeline of all paired samples")
plt.show()
```

```python
month_counts = pd.Series(dates).dt.month.value_counts().sort_index()

plt.figure(figsize=(8, 4))
plt.bar(month_counts.index, month_counts.values)
plt.xticks(range(1, 13))
plt.xlabel("Month")
plt.ylabel("Number of pairs")
plt.title("Seasonal distribution (all years combined)")
plt.show()
```

```python
df_time = pd.DataFrame({"date": dates})
df_time["year"] = df_time["date"].dt.year
df_time["month"] = df_time["date"].dt.month

heat = df_time.groupby(["year", "month"]).size().unstack(fill_value=0)
```

```python
# pick bins that make sense for your density
levels = [0, 1, 5, 10, 20, 50, 100, 200, 300, 400]
cmap = plt.cm.viridis
norm = BoundaryNorm(levels, cmap.N, clip=True)

plt.figure(figsize=(12, 6))
im = plt.imshow(heat, aspect="auto", cmap=cmap, norm=norm)

cbar = plt.colorbar(im, ticks=levels)
cbar.set_label("Pairs")

plt.yticks(range(len(heat.index)), heat.index)
plt.xticks(range(12), range(1, 13))
plt.xlabel("Month")
plt.ylabel("Year")
plt.title("Year–Month coverage of all paired samples (discrete scale)")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df_time = pd.DataFrame({"date": dates})
df_time["year"] = df_time["date"].dt.year
df_time["month"] = df_time["date"].dt.month

train = df_time[df_time["year"].between(2014, 2019)]
val   = df_time[df_time["year"] == 2020]
test  = df_time[df_time["year"] == 2021]

def plot_month_dist(df, title):
    month_counts = df["month"].value_counts().sort_index()
    n = len(df)

    plt.bar(month_counts.index, month_counts.values)
    plt.xticks(range(1, 13))
    plt.xlabel("Month")
    plt.ylabel("Number of pairs")
    plt.title(f"{title}  (N = {n})")

```

```python
REGION = "region-27_0-82_951-35_52-83_8" 

# extract region + time from index
regions = [k[0] for k in paired.index]
dates   = pd.to_datetime([k[1] for k in paired.index], errors="coerce")

df_time = pd.DataFrame({
    "region": regions,
    "date": dates
}).dropna()

df_time["year"] = df_time["date"].dt.year
df_time["month"] = df_time["date"].dt.month

# filter to one region
df_r = df_time[df_time["region"] == REGION]

train_years = [2014, 2015, 2016, 2017, 2018, 2019]

train = df_r[df_r["year"].isin(train_years)]
val   = df_r[df_r["year"] == 2020]
test  = df_r[df_r["year"] == 2021]

import matplotlib.pyplot as plt

def plot_month_dist(df, title):
    month_counts = df["month"].value_counts().sort_index()
    n = len(df)

    plt.bar(month_counts.index, month_counts.values)
    plt.xticks(range(1, 13))
    plt.xlabel("Month")
    plt.ylabel("Number of pairs")
    plt.title(f"{title}  (N = {n})")
```

```python
plt.figure(figsize=(10, 8))

plt.subplot(3, 1, 1)
plot_month_dist(train, f"Train (2014-2019) — {REGION}")

plt.subplot(3, 1, 2)
plot_month_dist(val, f"Validation (2020) — {REGION}")

plt.subplot(3, 1, 3)
plot_month_dist(test, f"Test (2021) — {REGION}")

plt.tight_layout()
plt.show()
```

```python
THRESH_FT = 0
THRESH_PM = 400
# mask = (
#     (paired["past_n_pm_good"] >= THRESH) &
#     (paired["future_n_pm_good"] >= THRESH)
# )

mask = (
    (paired["past_n_ft"] >= THRESH_FT) &
    (paired["future_n_ft"] >= THRESH_FT) &
    (paired["past_n_pm_good"] >= THRESH_PM) &
    (paired["future_n_pm_good"] >= THRESH_PM)
)

paired_thr = paired[mask]

print("Pairs before threshold:", len(paired))
print(f"Pairs after threshold (n_ft ≥ {THRESH_FT}, n_pm_good ≥ {THRESH_PM}):", len(paired_thr))
```

```python
dates_thr = pd.to_datetime(
    [k[1] for k in paired_thr.index],
    errors="coerce"
)
dates_thr = dates_thr[~pd.isna(dates_thr)]
```

```python
plt.figure(figsize=(12, 3))
plt.plot(dates_thr, [1]*len(dates_thr), "|")
plt.yticks([])
plt.xlabel("Date")
plt.title(f"Timeline of paired samples (n_ft ≥ {THRESH_FT}, n_pm_good ≥ {THRESH_PM}, both past & future)")
plt.show()
```

```python
month_counts_thr = pd.Series(dates_thr).dt.month.value_counts().sort_index()

plt.figure(figsize=(8, 4))
plt.bar(month_counts_thr.index, month_counts_thr.values)
plt.xticks(range(1, 13))
plt.xlabel("Month")
plt.ylabel("Number of pairs")
plt.title(f"Seasonal distribution (n_ft ≥ {THRESH_FT}, n_pm_good ≥ {THRESH_PM})")
plt.show()
```

```python
df_time_thr = pd.DataFrame({"date": dates_thr})
df_time_thr["year"] = df_time_thr["date"].dt.year
df_time_thr["month"] = df_time_thr["date"].dt.month

heat_thr = df_time_thr.groupby(["year", "month"]).size().unstack(fill_value=0)

levels = [0, 1, 5, 10, 20, 50, 100, 200, 300, 400]
cmap = plt.cm.viridis
norm = BoundaryNorm(levels, cmap.N, clip=True)

plt.figure(figsize=(12, 6))
im = plt.imshow(heat_thr, aspect="auto", cmap=cmap, norm=norm)

cbar = plt.colorbar(im, ticks=levels)
cbar.set_label("Pairs")

plt.yticks(range(len(heat_thr.index)), heat_thr.index)
plt.xticks(range(12), range(1, 13))
plt.xlabel("Month")
plt.ylabel("Year")
plt.title(f"Year–Month coverage (n_ft ≥ {THRESH_FT}, n_pm_good ≥ {THRESH_PM}, both sides)")
plt.show()

# Total per year
year_totals = heat_thr.sum(axis=1)

print("Total pairs per year:")
print(year_totals)
```
