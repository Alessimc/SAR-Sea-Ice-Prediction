```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns

TEXTWIDTH_PT = 418.25368
TEXTWIDTH_IN = TEXTWIDTH_PT / 72.27

def setup_pub_style(fontsize=9):
    mpl.rcParams.update({
        "font.size": fontsize,
        "axes.titlesize": fontsize,
        "axes.labelsize": fontsize,
        "xtick.labelsize": fontsize - 1,
        "ytick.labelsize": fontsize - 1,
        "legend.fontsize": fontsize - 1,
        "figure.dpi": 300,
        "savefig.dpi": 300,
    })

def fig_textwidth(height_ratio=0.62):
    return plt.subplots(figsize=(TEXTWIDTH_IN, TEXTWIDTH_IN * height_ratio))

# call once in your script/notebook
setup_pub_style(fontsize=9)
sns.set_theme(style="whitegrid", context="paper")  # seaborn styling
sns.set_palette("colorblind")                     # colorblind-safe palette


def plot_losses_7(
    csv1, csv2, csv3, csv4, csv5, csv6, csv7,
    labels=None,
    persistence_baseline=7.739443e-01,
    wind_baseline_30deg=1.619641e-01,
    title="",
    yscale="log",
    height_ratio=0.62,
    savefig=None
):
    """
    Plot train/val loss from 7 runs on the same axes.

    Each CSV must contain columns: epoch, train_loss, val_loss

    Convention:
      - train = solid
      - val   = dashed
    """
    csvs = [csv1, csv2, csv3, csv4, csv5, csv6, csv7]
    if labels is None:
        labels = [f"run{i+1}" for i in range(7)]
    if len(labels) != 7:
        raise ValueError("labels must have length 7")

    fig, ax = fig_textwidth(height_ratio=height_ratio)

    mins = []
    for path, lab in zip(csvs, labels):
        df = pd.read_csv(path).sort_values("epoch")

        # train/val (uncomment train if you want it)
        # ax.plot(df["epoch"], df["train_loss"], linewidth=1.6, linestyle="-",  label=f"{lab} train")
        ax.plot(df["epoch"], df["val_loss"],   linewidth=1.6, label=f"{lab}")

        mins.append((
            lab,
            float(np.nanmin(df["train_loss"])),
            float(np.nanmin(df["val_loss"]))
        ))

    # baselines
    ax.axhline(persistence_baseline, linestyle="--", linewidth=1.2, label="Persistence baseline", color="black")
    ax.axhline(wind_baseline_30deg,  linestyle=":", linewidth=1.2, label="2% wind + 30° rotation baseline", color="black")

    ax.set_xlabel("Epoch")
    ax.set_ylabel("MSE loss")
    ax.set_title(title)
    ax.set_xlim(left=1)

    if yscale:
        ax.set_yscale(yscale)
    import matplotlib.ticker as mticker

    if yscale == "log":
        ax.set_yscale("log")

        # Fix visible range explicitly
        ax.set_ylim(1e-1, 1.2e0)

        # Major ticks at 10^-1 and 10^0
        ax.yaxis.set_major_locator(mticker.LogLocator(base=10))
        ax.yaxis.set_major_formatter(mticker.LogFormatterMathtext(base=10))

        # Minor ticks at 2–9 × 10^-1
        ax.yaxis.set_minor_locator(
            mticker.LogLocator(base=10, subs=np.arange(2, 10) * 0.1)
        )
        ax.yaxis.set_minor_formatter(mticker.NullFormatter())

    ax.legend(ncol=2, frameon=True)

    fig.tight_layout()
    if savefig:
        plt.savefig(savefig, bbox_inches="tight", dpi=300)

    plt.show()

    for lab, tr_min, va_min in mins:
        print(f"{lab:>10s}  min train: {tr_min:.6g}   min val: {va_min:.6g}")
```
