```python
import pandas as pd
import seaborn as sns

def plot_distribution_by_season(df, param, pol, save=None):
    df = df.copy()

    df["date"] = pd.to_datetime(
        df["npz_path"].str.extract(r"/(\d{4})/(\d{2})/(\d{2})/").agg("-".join, axis=1),
        format="%Y-%m-%d",
        errors="coerce",
    )

    def month_to_season(m):
        return (
            "DJF" if m in (12, 1, 2) else
            "MAM" if m in (3, 4, 5) else
            "JJA" if m in (6, 7, 8) else
            "SON"
        )

    df["season"] = df["date"].dt.month.map(month_to_season)
    df["season"] = pd.Categorical(
        df["season"],
        categories=["DJF", "MAM", "JJA", "SON"],
        ordered=True
    )

    param_name = {
        "n_ft": "# of matched features",
        "n_pm_good": "# of matched patterns",
    }.get(param, param)

    fig, ax = fig_textwidth(height_ratio=0.55)

    sns.boxplot(
        data=df,
        x="season",
        y=param,
        palette="colorblind",
        ax=ax,
        width=0.6,
        fliersize=3,   # outlier size
        linewidth=1
    )


    # Add counts
    counts = df["season"].value_counts().sort_index()
    y_max = df[param].max()

    for i, season in enumerate(counts.index):
        ax.text(
            i,
            y_max * 0.995,
            f"n={counts[season]}",
            ha="center",
            va="bottom",
            fontweight="bold"
        )

    ax.set_title(f"{param_name} by season for {pol}-pol")
    ax.set_xlabel("Season")
    ax.set_ylabel(param_name)

    fig.tight_layout()

    if save:
        fig.savefig(save, bbox_inches="tight")

    return fig, ax
```

```python
import pandas as pd
import seaborn as sns

def _add_season(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["date"] = pd.to_datetime(
        df["npz_path"].str.extract(r"/(\d{4})/(\d{2})/(\d{2})/").agg("-".join, axis=1),
        format="%Y-%m-%d",
        errors="coerce",
    )

    def month_to_season(m):
        return (
            "DJF" if m in (12, 1, 2) else
            "MAM" if m in (3, 4, 5) else
            "JJA" if m in (6, 7, 8) else
            "SON"
        )

    df["season"] = df["date"].dt.month.map(month_to_season)
    df["season"] = pd.Categorical(df["season"], categories=["DJF", "MAM", "JJA", "SON"], ordered=True)
    return df


def plot_season_boxplot_compare_pol(df_hh: pd.DataFrame,
                                   df_hv: pd.DataFrame,
                                   param: str,
                                   save: str = None):
    """
    Compare HH vs HV distributions by season using side-by-side boxplots.

    Expects both dfs to have:
      - column 'npz_path' with /YYYY/MM/DD/ somewhere in path
      - column `param` to plot
    """
    hh = _add_season(df_hh)
    hv = _add_season(df_hv)
    hh["pol"] = "HH"
    hv["pol"] = "HV"

    df = pd.concat([hh, hv], ignore_index=True)
    df = df.dropna(subset=["season", param, "pol"])

    param_name = {
        "n_ft": "# of matched features",
        "n_pm_good": "# of matched patterns",
    }.get(param, param)

    fig, ax = fig_textwidth(height_ratio=0.55)

    sns.boxplot(
        data=df,
        x="season",
        y=param,
        hue="pol",
        palette="colorblind",
        ax=ax,
        width=0.7,
        fliersize=2.5,
        linewidth=1,
        dodge=True,
    )

    # Counts: show HH/HV counts per season above the boxes
    y_max = df[param].max()
    counts = df.groupby(["season", "pol"]).size().unstack(fill_value=0)

    for i, season in enumerate(df["season"].cat.categories):
        if season not in counts.index:
            continue
        n_hh = int(counts.loc[season].get("HH", 0))
        n_hv = int(counts.loc[season].get("HV", 0))
        ax.text(i, y_max * 0.995, f"HH={n_hh}  HV={n_hv}", ha="center", va="bottom", fontweight="bold")

    ax.set_title(f"{param_name} by season (HH vs HV)")
    ax.set_xlabel("Season")
    ax.set_ylabel(param_name)

    ax.legend(title="Pol", frameon=False, loc="upper right")

    fig.tight_layout()
    if save:
        fig.savefig(save, bbox_inches="tight")
    return fig, ax


fig, ax = plot_season_boxplot_compare_pol(df_HH, df_HV, param="n_ft", save="n_ft_season_hh_vs_hv.pdf")
plt.show()

fig, ax = plot_season_boxplot_compare_pol(df_HH, df_HV, param="n_pm_good", save="n_pm_good_season_hh_vs_hv.pdf")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def _add_season(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df["date"] = pd.to_datetime(
        df["npz_path"].str.extract(r"/(\d{4})/(\d{2})/(\d{2})/").agg("-".join, axis=1),
        format="%Y-%m-%d",
        errors="coerce",
    )

    def month_to_season(m):
        return (
            "DJF" if m in (12, 1, 2) else
            "MAM" if m in (3, 4, 5) else
            "JJA" if m in (6, 7, 8) else
            "SON"
        )

    df["season"] = df["date"].dt.month.map(month_to_season)
    df["season"] = pd.Categorical(df["season"], categories=["DJF", "MAM", "JJA", "SON"], ordered=True)
    return df


def plot_season_boxplot_compare_3(
    df1: pd.DataFrame, label1: str,
    df2: pd.DataFrame, label2: str,
    df3: pd.DataFrame, label3: str,
    param: str,
    save: str = None,
    height_ratio: float = 0.55,
):
    a = _add_season(df1); a["group"] = label1
    b = _add_season(df2); b["group"] = label2
    c = _add_season(df3); c["group"] = label3

    df = pd.concat([a, b, c], ignore_index=True).dropna(subset=["season", param, "group"])

    param_name = {
        "n_ft": "# of matched features",
        "n_pm_good": "# of matched patterns",
    }.get(param, param)

    fig, ax = fig_textwidth(height_ratio)

    sns.boxplot(
        data=df,
        x="season",
        y=param,
        hue="group",
        palette="colorblind",
        ax=ax,
        width=0.75,
        fliersize=2.5,
        linewidth=1,
        dodge=True,
    )

    # ---- counts table ----
    counts = (
        df.groupby(["group", "season"])
          .size()
          .unstack("season", fill_value=0)
          .reindex(index=[label1, label2, label3], columns=df["season"].cat.categories)
    )

    table = ax.table(
        cellText=counts.values,
        rowLabels=[f"{g} (n)" for g in counts.index],
        colLabels=counts.columns,
        cellLoc="center",
        rowLoc="center",
        loc="bottom",
        bbox=[0.0, -0.32, 1.0, 0.30],
    )

    

    table.auto_set_font_size(False)
    table.set_fontsize(8)

    for (r, c), cell in table.get_celld().items():
        cell.set_linewidth(0.5)
        cell.set_edgecolor("0.5")
        if r == 0:
            cell.visible_edges = "BLR"

    # ---- Axis adjustments ----
    ax.set_xlabel("")                 # remove axis label
    ax.set_xticklabels([])            # remove DJF/MAM/JJA/SON tick labels
    ax.set_ylabel(param_name)         # keep y-axis label

    # ---- Legend styling ----
    leg = ax.legend(
        loc="upper left",
        frameon=True,
        facecolor="white",
        edgecolor="0.7",
        framealpha=0.95,
    )
    leg.set_title(None)

    sns.despine(ax=ax)

    plt.subplots_adjust(bottom=0.28)

    fig.tight_layout()

    if save:
        fig.savefig(save, bbox_inches="tight")

    return fig, ax
```

```python
import pandas as pd

def compute_correlation(df: pd.DataFrame):
    # drop missing values in either column
    df = df.dropna(subset=["n_pm_good", "n_ft"])

    # Pearson correlation
    corr = df["n_pm_good"].corr(df["n_ft"], method="pearson")

    corr_s = df["n_pm_good"].corr(df["n_ft"], method="spearman")
    print(f"Spearman correlation: {corr_s:.3f}")

    print(f"Pearson correlation (n_pm_good vs n_ft): {corr:.3f}")
    return corr
```
