```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import matplotlib.ticker as mticker
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import numpy as np

# ---------------- Publication style ----------------
TEXTWIDTH_PT = 418.25368
TEXTWIDTH_IN = TEXTWIDTH_PT / 72.27

def setup_pub_style(fontsize=9):
    mpl.rcParams.update({
        "font.size": fontsize,
        "axes.titlesize": fontsize,
        "axes.labelsize": fontsize,
        "xtick.labelsize": fontsize - 1,
        "ytick.labelsize": fontsize - 1,
        "legend.fontsize": fontsize - 1,
        "figure.dpi": 300,
        "savefig.dpi": 300,
    })

setup_pub_style(fontsize=9)

# ---------------- Regions ----------------
regions = [
    "region-27_0-82_951-35_52-83_8",      # 6
    "region-27_2-82_348-35_0-83_2",       # 5
    "region-27_765-81_753-35_0-82_6",     # 10
    "region-33_0-82_95-40_715-83_903",    # 7
    "region-33_07-82_34-40_1-83_297",     # 8
    "region-33_2-81_641-39_585-82_6",     # 9
    "region-38_38-81_728-45_44-82_6",     # 13
    "region-38_65-82_33-45_61-83_293",    # 12
    "region-38_38-82_95-46_0-83_913",
]

region_labels = ["6", "5", "10", "7", "9", "8", "13", "11", "12"]

def _parse_region_box(region_str):
    s = region_str.replace("region-", "")
    parts = s.split("-")
    if len(parts) != 4:
        raise ValueError(f"Unexpected region format: {region_str}")

    lon1 = float(parts[0].replace("_", "."))
    lat1 = float(parts[1].replace("_", "."))
    lon2 = float(parts[2].replace("_", "."))
    lat2 = float(parts[3].replace("_", "."))
    return lon1, lat1, lon2, lat2

def _fmt_lon(lon):
    if lon < 0:
        return f"{abs(int(lon))}°W"
    elif lon > 0:
        return f"{int(lon)}°E"
    else:
        return "0°"

def plot_region_boxes_only(
    region_list,
    labels=None,
    title="Study area boxes northeast of Svalbard",
    save="study_area_boxes_only.pdf",
    height_ratio=0.70,
    extent=None,                 # [lon_min, lon_max, lat_min, lat_max]
    use_stock_img=True,
    ocean_color="#b7d7ee",
    land_color="0.90",
    label_boxes=True,
    meridians=np.arange(-10, 56, 5),
    parallels=np.arange(76, 86, 2),
    meridian_label_lat=75.6,     # where lon labels go
    parallel_label_lon=-7.0,     # where lat labels go
):
    boxes = [_parse_region_box(r) for r in region_list]

    box_colors = ["firebrick" if i % 2 == 0 else "lightcoral" for i in range(len(boxes))]

    proj = ccrs.NorthPolarStereo()
    pc = ccrs.PlateCarree()

    fig = plt.figure(figsize=(TEXTWIDTH_IN, TEXTWIDTH_IN * height_ratio), dpi=300)
    ax = fig.add_subplot(1, 1, 1, projection=proj)

    if extent is not None:
        ax.set_extent(extent, crs=pc)

    # Background
    if use_stock_img:
        ax.stock_img()
    else:
        ax.set_facecolor(ocean_color)
        ax.add_feature(cfeature.LAND, facecolor=land_color, edgecolor="none", zorder=0)

    ax.coastlines(color="black", linewidth=0.8, zorder=2)

    # Gridlines only, no auto labels
    gl = ax.gridlines(
        crs=pc,
        draw_labels=False,
        linewidth=0.5,
        color="0.35",
        alpha=0.55,
        linestyle="--",
        x_inline=False,
        y_inline=False,
    )
    gl.xlocator = mticker.FixedLocator(meridians)
    gl.ylocator = mticker.FixedLocator(parallels)

    # Boxes
    for i, (lon1, lat1, lon2, lat2) in enumerate(boxes):
        color = box_colors[i]

        xs_box = [lon1, lon2, lon2, lon1, lon1]
        ys_box = [lat1, lat1, lat2, lat2, lat1]

        ax.fill(
            xs_box, ys_box,
            transform=pc,
            facecolor=color,
            alpha=0.16,
            edgecolor="none",
            zorder=3,
        )

        ax.plot(
            xs_box, ys_box,
            transform=pc,
            color=color,
            linewidth=1.4,
            alpha=0.95,
            zorder=4,
        )

    # ---- Manual longitude labels inside figure ----
    for lon in meridians:
        ax.text(
            lon, meridian_label_lat,
            _fmt_lon(lon),
            transform=pc,
            ha="center",
            va="bottom",
            fontsize=8,
            color="black",
            zorder=10,
            bbox=dict(
                boxstyle="round,pad=0.12",
                facecolor="none",
                edgecolor="none",
                alpha=0.55,
            ),
        )

    # ---- Manual latitude labels inside figure ----
    for lat in parallels:
        ax.text(
            parallel_label_lon, lat,
            f"{int(lat)}°N",
            transform=pc,
            ha="left",
            va="center",
            fontsize=8,
            color="black",
            zorder=10,
            bbox=dict(
                boxstyle="round,pad=0.12",
                facecolor="none",
                edgecolor="none",
                alpha=0.55,
            ),
        )

    ax.set_title(title)
    fig.subplots_adjust(left=0.02, right=0.98, top=0.93, bottom=0.05)

    if save:
        fig.savefig(save, bbox_inches="tight", dpi=300)

    plt.show()
    return fig, ax


# ---- Usage ----
plot_region_boxes_only(
    regions,
    labels=region_labels,
    title="",
    save="study_area_boxes_only.pdf",
    height_ratio=0.70,
    extent=[-10, 55, 75.0, 85],
    use_stock_img=False,
    meridians=np.arange(0, 56, 10),
    parallels=np.arange(76, 86, 2),
    meridian_label_lat=75.6,
    parallel_label_lon=-7.0,
)
```
