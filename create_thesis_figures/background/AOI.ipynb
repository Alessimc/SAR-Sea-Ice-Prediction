```python
import os
from datetime import datetime

import numpy as np
import xarray as xr
import pyproj
import matplotlib.pyplot as plt
import matplotlib as mpl
import matplotlib.ticker as mticker
from matplotlib.lines import Line2D
from scipy import ndimage as ndi
from skimage.measure import find_contours

import cartopy.crs as ccrs
import cartopy.feature as cfeature


TEXTWIDTH_IN = 418.25368 / 72.27
mpl.rcParams.update({
    "font.size": 9,
    "figure.dpi": 300,
    "savefig.dpi": 300,
})


CARRA_FILE = ""

AROME_URL = (
    "https://thredds.met.no/thredds/dodsC/aromearcticarchive"
    "/2020/06/24/arome_arctic_vtk_20200624T18Z.nc"
)


SIC_ROOT = ""
SIC_GRID_NC = os.path.join(SIC_ROOT, "Regrid_latlon.nc")
SIC_PATH_TEMPLATE = "{root}/{yyyy}/{mm}/RegridAMSR2_{yyyymmdd}.nc"

SIC_VAR = "amsr2"
SIC_LAT_VAR = "Latitude"
SIC_LON_VAR = "Longitude"

YEARS = list(range(2014, 2022))   # 2014–2021 inclusive
SIC_THRESHOLD = 0.15              # 15% SIC threshold
MEDIAN_CONTOUR_LEVEL = 0.5        # 50% probability contour = median ice edge


# ── Easy styling control ───────────────────────────────────────────────
STYLE = {
    "study_area": {
        "line_color": "firebrick",
        "fill_color": "white",
        "lw": 1.4,
        "ls": "-",
        "fill_alpha": 0.15,
    },
    "carra": {
        "line_color": "#2ca02c",
        "fill_color": "white",
        "lw": 1.4,
        "ls": "-",
        "fill_alpha": 0.15,
    },
    "march_edge": {
        "line_color": "black",
        "lw": 0.8,
        "ls": "--",
    },
    "sept_edge": {
        "line_color": "firebrick",
        "lw": 0.8,
        "ls": "--",
    },
    "arome": {
        "line_color": "#1f77b4",
        "fill_color": "white",
        "lw": 1.4,
        "ls": "-",
        "fill_alpha": 0.15,
    },
}


# ── Helper functions ───────────────────────────────────────────────────
def domain_boundary_lonlat(x, y, crs, n_pts=200):
    x0, x1 = x.min(), x.max()
    y0, y1 = y.min(), y.max()

    xs = np.concatenate([
        np.linspace(x0, x1, n_pts),
        np.full(n_pts, x1),
        np.linspace(x1, x0, n_pts),
        np.full(n_pts, x0),
    ])

    ys = np.concatenate([
        np.full(n_pts, y0),
        np.linspace(y0, y1, n_pts),
        np.full(n_pts, y1),
        np.linspace(y1, y0, n_pts),
    ])

    transformer = pyproj.Transformer.from_crs(crs, "EPSG:4326", always_xy=True)
    lons, lats = transformer.transform(xs, ys)
    return lons, lats


def _parse_region_box(region_str):
    parts = region_str.replace("region-", "").split("-")
    return tuple(float(p.replace("_", ".")) for p in parts)


def extract_2d_array(ds: xr.Dataset, var_name: str, path: str) -> np.ndarray:
    if var_name not in ds.variables:
        raise KeyError(
            f"Variable {var_name!r} not found in {path}. "
            f"Available variables: {list(ds.variables)}"
        )

    arr = ds[var_name].values

    if arr.ndim == 3 and arr.shape[0] == 1:
        arr = arr[0]

    if arr.ndim != 2:
        raise ValueError(
            f"Variable {var_name!r} in {path} must be 2D or single-band 3D; "
            f"got shape {arr.shape}"
        )

    return np.asarray(arr)


def amsr2_path_for_date(year: int, month: int, day: int) -> str:
    dt = datetime(year, month, day)
    return SIC_PATH_TEMPLATE.format(
        root=SIC_ROOT,
        yyyy=dt.strftime("%Y"),
        yy=dt.strftime("%y"),
        mm=dt.strftime("%m"),
        dd=dt.strftime("%d"),
        yyyymmdd=dt.strftime("%Y%m%d"),
    )


def load_amsr2_grid():
    with xr.open_dataset(SIC_GRID_NC) as ds_grid:
        lat = extract_2d_array(ds_grid, SIC_LAT_VAR, SIC_GRID_NC).astype(np.float64, copy=False)
        lon = extract_2d_array(ds_grid, SIC_LON_VAR, SIC_GRID_NC).astype(np.float64, copy=False)
    return lon, lat


def compute_probability_map(month: int, day: int, years, threshold=0.15):
    binary_maps = []
    used_years = []
    missing_years = []

    for year in years:
        path = amsr2_path_for_date(year, month, day)

        if not os.path.exists(path):
            missing_years.append(year)
            continue

        with xr.open_dataset(path) as ds:
            sic = extract_2d_array(ds, SIC_VAR, path).astype(np.float64, copy=False)

        valid = np.isfinite(sic) & (sic >= 0.0) & (sic <= 1.0)

        binary = np.full(sic.shape, np.nan, dtype=np.float32)
        binary[valid] = (sic[valid] > threshold).astype(np.float32)

        binary_maps.append(binary)
        used_years.append(year)

    if not binary_maps:
        raise RuntimeError(f"No AMSR2 files found for {month:02d}-{day:02d}")

    prob = np.nanmean(np.stack(binary_maps, axis=0), axis=0)

    print(f"{month:02d}-{day:02d}: used years = {used_years}")
    if missing_years:
        print(f"{month:02d}-{day:02d}: missing years = {missing_years}")

    return prob, used_years


def clean_main_ice_mask(prob_map, contour_level=0.5, closing_iters=1, opening_iters=1):
    mask = np.isfinite(prob_map) & (prob_map >= contour_level)

    structure = ndi.generate_binary_structure(2, 2)

    if closing_iters > 0:
        mask = ndi.binary_closing(mask, structure=structure, iterations=closing_iters)

    if opening_iters > 0:
        mask = ndi.binary_opening(mask, structure=structure, iterations=opening_iters)

    mask = ndi.binary_fill_holes(mask)

    labels, n_labels = ndi.label(mask, structure=structure)

    if n_labels == 0:
        return np.zeros_like(mask, dtype=bool)

    counts = np.bincount(labels.ravel())
    counts[0] = 0
    largest_label = counts.argmax()

    return labels == largest_label


def mask_to_contour_field(mask):
    return np.where(mask, 1.0, 0.0).astype(np.float32)


def extract_open_contour_segments(mask_field, lon2d, lat2d, level=0.5, min_points=30):
    contours_rc = find_contours(mask_field, level=level)
    segments = []

    for rc in contours_rc:
        if rc.shape[0] < min_points:
            continue

        rows = rc[:, 0]
        cols = rc[:, 1]

        lons = ndi.map_coordinates(lon2d, [rows, cols], order=1, mode="nearest")
        lats = ndi.map_coordinates(lat2d, [rows, cols], order=1, mode="nearest")

        good = np.isfinite(lons) & np.isfinite(lats)
        if np.count_nonzero(good) < min_points:
            continue

        segments.append((lons[good], lats[good]))

    return segments


def plot_contour_segments(ax, segments, color, linewidth, linestyle, transform, zorder):
    for lons, lats in segments:
        ax.plot(
            lons,
            lats,
            color=color,
            linewidth=linewidth,
            linestyle=linestyle,
            transform=transform,
            zorder=zorder,
        )


def plot_domain(ax, lons, lats, style, transform, zfill, zline):
    ax.fill(
        lons,
        lats,
        transform=transform,
        facecolor=style["fill_color"],
        alpha=style["fill_alpha"],
        edgecolor="none",
        zorder=zfill,
    )

    ax.plot(
        lons,
        lats,
        transform=transform,
        color=style["line_color"],
        linewidth=style["lw"],
        linestyle=style["ls"],
        zorder=zline,
    )


def fmt_lon_label(lon):
    if lon < 0:
        return f"{abs(int(lon))}°W"
    if lon > 0:
        return f"{int(lon)}°E"
    return "0°"


def fmt_lat_label(lat):
    return f"{int(lat)}°N"


def add_internal_lon_labels(ax, lons, lat, transform, **text_kwargs):
    for lon in lons:
        ax.text(
            lon,
            lat,
            fmt_lon_label(lon),
            transform=transform,
            ha="center",
            va="center",
            **text_kwargs,
        )


def add_internal_lat_labels(ax, lats, lon, transform, **text_kwargs):
    for lat in lats:
        ax.text(
            lon,
            lat,
            fmt_lat_label(lat),
            transform=transform,
            ha="center",
            va="center",
            **text_kwargs,
        )


ds_carra = xr.open_dataset(CARRA_FILE)

lcc = ds_carra["Lambert_Conformal"]
carra_crs = pyproj.CRS.from_proj4(
    f"+proj=lcc "
    f"+lat_1={float(lcc.standard_parallel)} +lat_2={float(lcc.standard_parallel)} "
    f"+lat_0={float(lcc.latitude_of_projection_origin)} "
    f"+lon_0={float(lcc.longitude_of_central_meridian)} "
    f"+x_0={float(lcc.false_easting)} +y_0={float(lcc.false_northing)} "
    f"+R={float(lcc.earth_radius)} +units=m +no_defs"
)

carra_lons, carra_lats = domain_boundary_lonlat(
    ds_carra["x"].values,
    ds_carra["y"].values,
    carra_crs,
)
ds_carra.close()


ds_arome = xr.open_dataset(AROME_URL)

arome_crs = pyproj.CRS.from_cf(ds_arome["projection_lambert"].attrs)
x_name = "x" if "x" in ds_arome.coords else "X"
y_name = "y" if "y" in ds_arome.coords else "Y"

arome_lons, arome_lats = domain_boundary_lonlat(
    ds_arome[x_name].values,
    ds_arome[y_name].values,
    arome_crs,
)
ds_arome.close()


regions = [
    "region-27_0-82_951-35_52-83_8",
    "region-27_2-82_348-35_0-83_2",
    "region-27_765-81_753-35_0-82_6",
    "region-33_0-82_95-40_715-83_903",
    "region-33_07-82_34-40_1-83_297",
    "region-33_2-81_641-39_585-82_6",
    "region-38_38-81_728-45_44-82_6",
    "region-38_65-82_33-45_61-83_293",
    "region-38_38-82_95-46_0-83_913",
]

boxes = [_parse_region_box(r) for r in regions]
lon_min = min(b[0] for b in boxes)
lat_min = min(b[1] for b in boxes)
lon_max = max(b[2] for b in boxes)
lat_max = max(b[3] for b in boxes)

xs_sa = [lon_min, lon_max, lon_max, lon_min, lon_min]
ys_sa = [lat_min, lat_min, lat_max, lat_max, lat_min]


sic_lon, sic_lat = load_amsr2_grid()

march_prob, march_years = compute_probability_map(
    month=3,
    day=15,
    years=YEARS,
    threshold=SIC_THRESHOLD,
)

sept_prob, sept_years = compute_probability_map(
    month=9,
    day=15,
    years=YEARS,
    threshold=SIC_THRESHOLD,
)

march_main_mask = clean_main_ice_mask(
    march_prob,
    contour_level=MEDIAN_CONTOUR_LEVEL,
    closing_iters=1,
    opening_iters=1,
)

sept_main_mask = clean_main_ice_mask(
    sept_prob,
    contour_level=MEDIAN_CONTOUR_LEVEL,
    closing_iters=1,
    opening_iters=1,
)

march_contour_field = mask_to_contour_field(march_main_mask)
sept_contour_field = mask_to_contour_field(sept_main_mask)

march_segments = extract_open_contour_segments(
    march_contour_field,
    sic_lon,
    sic_lat,
    level=0.5,
    min_points=30,
)

sept_segments = extract_open_contour_segments(
    sept_contour_field,
    sic_lon,
    sic_lat,
    level=0.5,
    min_points=30,
)

print(f"March contour segments kept: {len(march_segments)}")
print(f"September contour segments kept: {len(sept_segments)}")


proj = ccrs.NorthPolarStereo()
pc = ccrs.PlateCarree()

fig = plt.figure(figsize=(TEXTWIDTH_IN, TEXTWIDTH_IN * 0.85), dpi=300)
ax = fig.add_subplot(1, 1, 1, projection=proj)
ax.set_extent([-30, 80, 60, 90], crs=pc)

ax.set_facecolor("#b7d7ee")
ax.add_feature(cfeature.LAND, facecolor="0.90", edgecolor="none", zorder=0)
ax.coastlines(color="black", linewidth=0.8, zorder=1)

gl = ax.gridlines(
    crs=pc,
    draw_labels=False,
    linewidth=0.5,
    color="0.35",
    alpha=0.55,
    linestyle="--",
)

gl.xlocator = mticker.FixedLocator(np.arange(-80, 81, 20))
gl.ylocator = mticker.FixedLocator(np.arange(30, 91, 10))


# Internal lon/lat labels, not border labels
label_style = dict(
    fontsize=7,
    color="black",
    zorder=50,
    bbox=dict(
        facecolor="white",
        edgecolor="none",
        alpha=0.0,
        pad=0.2,
    ),
)

add_internal_lon_labels(
    ax,
    lons=[-20, 0, 20, 40, 60, 80],
    lat=62,
    transform=pc,
    **label_style,
)

add_internal_lat_labels(
    ax,
    lats=[80],
    lon=-28,
    transform=pc,
    **label_style,
)

add_internal_lat_labels(
    ax,
    lats=[70],
    lon=-33,
    transform=pc,
    **label_style,
)


# 1) Study area
plot_domain(
    ax,
    xs_sa,
    ys_sa,
    STYLE["study_area"],
    transform=pc,
    zfill=2,
    zline=3,
)

# 2) CARRA
plot_domain(
    ax,
    carra_lons,
    carra_lats,
    STYLE["carra"],
    transform=pc,
    zfill=4,
    zline=5,
)

# 3) SIC contours
plot_contour_segments(
    ax,
    march_segments,
    color=STYLE["march_edge"]["line_color"],
    linewidth=STYLE["march_edge"]["lw"],
    linestyle=STYLE["march_edge"]["ls"],
    transform=pc,
    zorder=6,
)

plot_contour_segments(
    ax,
    sept_segments,
    color=STYLE["sept_edge"]["line_color"],
    linewidth=STYLE["sept_edge"]["lw"],
    linestyle=STYLE["sept_edge"]["ls"],
    transform=pc,
    zorder=6,
)

# 4) AROME-Arctic
plot_domain(
    ax,
    arome_lons,
    arome_lats,
    STYLE["arome"],
    transform=pc,
    zfill=7,
    zline=8,
)


legend_handles = [
    Line2D(
        [0], [0],
        color=STYLE["study_area"]["line_color"],
        lw=STYLE["study_area"]["lw"],
        ls=STYLE["study_area"]["ls"],
        label="Study area",
    ),
    Line2D(
        [0], [0],
        color=STYLE["carra"]["line_color"],
        lw=STYLE["carra"]["lw"],
        ls=STYLE["carra"]["ls"],
        label="CARRA-East",
    ),
    Line2D(
        [0], [0],
        color=STYLE["arome"]["line_color"],
        lw=STYLE["arome"]["lw"],
        ls=STYLE["arome"]["ls"],
        label="AROME-Arctic",
    ),
    Line2D(
        [0], [0],
        color=STYLE["march_edge"]["line_color"],
        lw=STYLE["march_edge"]["lw"],
        ls=STYLE["march_edge"]["ls"],
        label="Median winter sea ice edge",
    ),
    Line2D(
        [0], [0],
        color=STYLE["sept_edge"]["line_color"],
        lw=STYLE["sept_edge"]["lw"],
        ls=STYLE["sept_edge"]["ls"],
        label="Median summer sea ice edge",
    ),
]

leg = ax.legend(
    handles=legend_handles,
    loc="lower right",
    fontsize=8,
    framealpha=0.8,
)

leg.set_zorder(100)

fig.savefig("model_domains.pdf", bbox_inches="tight", dpi=300)
plt.show()
```
